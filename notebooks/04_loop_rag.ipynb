# Loop RAG (PDCA)

plan → do → check → act until convergence

```python
from dotenv import load_dotenv
load_dotenv()

from rag.registry import run

out = run('loop', 'What is LangGraph?')
print(out['answer'])
```

```python
import json
print(json.dumps(out['trace'], indent=2, default=str))
```

```python
# PDCA iteration trace — one row per loop
for h in out['trace']:
    if h.get('node') == 'check':
        d = h['deviation']
        print(f"iter={h['iter']} total={d['total']:.3f} dominant={d['dominant']} converged={d['converged']}")
```
